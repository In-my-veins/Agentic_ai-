```python
from dotenv import load_dotenv
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_chroma import Chroma

load_dotenv()

# 1. Create the embedding model
embeddings = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001"
)

# 2. Create the Chroma vector store
vector_store = Chroma(
    collection_name="news_articles",
    embedding_function=embeddings,
    persist_directory="./chroma_db"
)

# 3. Add documents
documents = [
    "Apple is leading in smartphone sales with strong iPhone sales.",
    "Tesla booked a minor profit of 1 billion dollars in Q2.",
    "Apple shares rose after strong iPhone sales.",
    "SpaceX got a NASA contract worth 10 billion dollars.",
    "NVIDIA is leading the AI chip market with its powerful GPUs.",
]

ids = ["id1", "id2", "id3", "id4", "id5"]

vector_store.add_texts(
    texts=documents,
    ids=ids
)

print("Documents added successfully!")
```

```python
results = vector_store.similarity_search(
    "Which company is leading the AI chip market?",
    k=2
)

for i, document in enumerate(results, 1):
    print(f"\nResult {i}:")
    print(document.page_content)
```
