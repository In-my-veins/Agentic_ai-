```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

llm  = ChatGroq(model="openai/gpt-oss-120b")


with open("blood_ork.txt","r") as f:
    blood_ork = f.read()
    print(blood_ork)

extraction_prompt = f"""
You are a medical data extraction assistant.
From the blood report below, extract all test values and classify each one as HIGH, LOW, or NORMAL based on 
the reference ranges provided in the report.

Format your response as: 
    -Test Name: Value | Status: HIGH/LOW/NORMAL | Reference: range
Blood Report:
{blood_ork}
"""

response = llm.invoke(extraction_prompt)
extracted_values = response.text
print("===EXTRACTED VALUES===")
print(extracted_values)
    
```

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

llm  = ChatGroq(model="openai/gpt-oss-120b")


with open("blood_ork.txt","r") as f:
    blood_ork = f.read()
    print(blood_ork)

extraction_prompt = f"""
You are a medical data extraction assistant.
From the blood report below, extract all test values and classify each one as HIGH, LOW, or NORMAL based on 
the reference ranges provided in the report.

Format your response as: 
    -Test Name: Value | Status: HIGH/LOW/NORMAL | Reference: range
Blood Report:
{blood_ork}
"""

response = llm.invoke(extraction_prompt)
extracted_values = response.text
print("===EXTRACTED VALUES===")
print(extracted_values)
    
```

```python
nutrition_prompt = f"""
You are a cautious nutritional guidance assistant.

Based on the blood test results below, provide general nutritional guidance
that may support the patient's health.

Requirements:
1. Focus primarily on test results classified as HIGH or LOW.
2. Explain briefly why each abnormal result may be relevant to nutrition.
3. Recommend general foods and eating patterns that may support healthier
   values.
4. Mention foods or eating patterns that may be reasonable to limit when
   relevant.
5. Do not diagnose any disease or medical condition.
6. Do not prescribe medication or a strict medical diet.
7. Do not claim that food alone can treat or cure a medical condition.
8. Clearly state that the guidance is general educational information and
   that the patient should discuss abnormal results with a qualified
   healthcare professional.
9. Keep the advice practical and easy for a patient to understand.

Blood test results:
{extracted_values}
"""

response = llm.invoke(nutrition_prompt)

nutrition_advice = response.content

print("=== NUTRITIONAL GUIDANCE ===")
print(nutrition_advice)
```
